```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# display all rows in df
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

### Helminth Eggs

```python
classes_names = ['Hymenolepis nana', 'Hymenolepis diminuta', 'Ancylostomatidae', 
                 'Enterobius vermicularis', 'Ascaris lumbricoides', 'Trichuris trichiura',
                 'Schistosoma mansoni', 'Taenia spp.', 'Impurities']
```

```python
table = {
    'perc': [],
    classes_names[0]: [],
    classes_names[1]: [],
    classes_names[2]: [],
    classes_names[3]: [],
    classes_names[4]: [],
    classes_names[5]: [],
    classes_names[6]: [],
    classes_names[7]: [],
    classes_names[8]: [],
    'split': [],
    'kappa': [],
    'accuracy': []
}
```

```python
parent_dir = '/home/isabela/Codes/LibIFT/Results/helminth-eggs/gflimBoFP_incremental_splits_1e2/'
for split_folder in os.listdir(parent_dir):
    if os.path.isdir(os.path.join(parent_dir, split_folder)) == False:
        continue
    for perc_folder in ['perc5', 'perc25', 'perc50', 'perc75', 'perc100']:
        filename = os.path.join(parent_dir, split_folder, perc_folder, "9-1-1-16-max-3-125_3_noSkip",
                                    "300-250spx_act-kernel_act-pool_act-stride",
                                    "notFixedSPX_MEAN_CONCAT_STD_SIZE_SORTFEAT_CLASS_DOMINANCE",
                                    "classification_layer1_test_1e2.txt")
        if os.path.isfile(filename):
            with open(filename, "r") as f:
                lines = f.readlines()
            for line in lines:
                if line.startswith("kappa:"):
                    values = line.split(', ')
                    table['kappa'].append(float(values[0].split('kappa: ')[1]))
                    table['accuracy'].append(float(values[1].split('accuracy: ')[1].strip()))
                elif line.startswith("- class 1:"):
                    table[classes_names[0]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 2:"):
                    table[classes_names[1]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 3:"):
                    table[classes_names[2]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 4:"):
                    table[classes_names[3]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 5:"):
                    table[classes_names[4]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 6:"):
                    table[classes_names[5]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 7:"):
                    table[classes_names[6]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 8:"):
                    table[classes_names[7]].append(float(line.split(' ')[3]))
                elif line.startswith("- class 9:"):
                    table[classes_names[8]].append(float(line.split(' ')[3]))
                #elif line.startswith("- nsamples:"):
                #    values = [l.strip().split(': ')[1] for l in line.split(', ')]
                #    table['nsamples'].append(int(values[0]))
                #    table['nfeats'].append(int(values[1]))
            table['split'].append(int(split_folder.split('t')[1]))
            table['perc'].append(int(perc_folder.split('c')[1]))
df = pd.DataFrame(table)
display(df)
                            
```

```python
# display results per training percentage grouping by split
grouped_df = df.copy()
all_keys = ['perc', classes_names[0], classes_names[1], 
            classes_names[2], classes_names[3], classes_names[4], 
            classes_names[5], classes_names[6], classes_names[7], 
            classes_names[8], 'kappa', 'accuracy']

for k in [classes_names[0], classes_names[1], classes_names[2], classes_names[3], classes_names[4], 
          classes_names[5], classes_names[6], classes_names[7], 
          classes_names[8], 'kappa', 'accuracy']:
    grouped_df[k+'_mean'] = grouped_df.groupby('perc')[k].transform('mean')
    grouped_df[k+'_std'] = grouped_df.groupby('perc')[k].transform('std')

grouped_df = grouped_df.drop_duplicates(subset=['perc']).reset_index(drop=True)

grouped_df = grouped_df[['perc'] + [k+'_mean' for k in all_keys if k not in ['perc']] + [k+'_std' for k in all_keys if k not in ['perc']]]
display(grouped_df.style.background_gradient(cmap='Blues'))
```

```python
# reorder columns to have perc first, then tuples of (mean, std) for each metric
ordered_cols = ['perc']
for k in [classes_names[0], classes_names[1], classes_names[2], classes_names[3], classes_names[4], 
          classes_names[5], classes_names[6], classes_names[7], 
          classes_names[8], 'kappa', 'accuracy']:
    ordered_cols.append(k+'_mean')
    ordered_cols.append(k+'_std')
grouped_df = grouped_df[ordered_cols]
display(grouped_df.style.background_gradient(cmap='Blues'))
```

```python
# transpose for better visualization
transposed_df = grouped_df.set_index('perc').T
display(transposed_df.style.background_gradient(cmap='Blues'))
```

```python
# put mean and std in the same cell for better visualization
for k in [classes_names[0], classes_names[1], classes_names[2], classes_names[3], classes_names[4], 
          classes_names[5], classes_names[6], classes_names[7], 
          classes_names[8], 'kappa', 'accuracy']:
    mean_row = transposed_df.loc[k+'_mean']
    std_row = transposed_df.loc[k+'_std']
    transposed_df.loc[k] = mean_row.combine(
        std_row, lambda m, s: f"{m:.4f} ± {s:.4f}"
    )
    transposed_df = transposed_df.drop(labels=[k+'_mean', k+'_std'])
display(transposed_df.style)
```

```python
# save as csv
transposed_df.to_csv('/home/isabela/Codes/LibIFT/Results/gflim_results_eggs.csv')
```
